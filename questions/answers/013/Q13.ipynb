```python
from pathlib import Path
import pandas as pd

# Define the path of your folder containing 60 zips
zip_folder = Path(r"C:\Users\ASUS\Desktop\tennis_data")

# Read proper tables
MatchAwayTeamInfo_df = pd.read_parquet(zip_folder / "away_team_all.parquet")

MatchHomeTeamInfo_df = pd.read_parquet(zip_folder / "home_team_all.parquet")

# Concat two tables to collect all players, then drop duplicates
players = pd.concat([ MatchHomeTeamInfo_df, MatchAwayTeamInfo_df], ignore_index=True)

players = players.drop_duplicates(subset="player_id")
```

```python
# Find number of left-handed and right-handed players
# NaN values will also be dropped
LeftHanded_df = players[players["plays"] == "left-handed"].shape[0]
print(LeftHanded_df)

RightHanded_df = players[players["plays"] == "right-handed"].shape[0]
print(RightHanded_df)
```
