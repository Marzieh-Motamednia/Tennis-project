```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import duckdb
```

```python
con = duckdb.connect(r'C:\Users\marzieh\Documents\GitHub\Tennis-project\data\tennis.duckdb', read_only=True)
con.execute("SHOW TABLES").df()
```

```python
query = """
    SELECT  period_1,
            period_2,
            period_3,
            period_4,
            period_5
    FROM match_time
    WHERE   period_1 IS NOT NULL AND
            match_id IS NOT NULL
    QUALIFY ROW_NUMBER() OVER (
        PARTITION BY match_id
        ORDER BY match_id
    ) = 1
"""
df = con.execute(query).df()
df
```

```python
cols = ["period_1", "period_2", "period_3", "period_4", "period_5"]

for col in cols:
    df[col] = df[col].where(df[col].isna() | (df[col] >= 300))

df
```
