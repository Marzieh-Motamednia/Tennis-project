```python
query1 = """
SELECT AVG(breaks_per_match) AS avg_breaks_of_serve_per_match
FROM (
    SELECT match_id, COUNT(*) AS breaks_per_match
    FROM power
    WHERE break_occurred = TRUE
    GROUP BY match_id
) AS match_breaks
"""

df1 = con.execute(query1).df()
df1
```
