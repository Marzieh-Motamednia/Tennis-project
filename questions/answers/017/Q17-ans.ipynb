```python
query = """
    SELECT AVG(break_games_count) AS avg_break_games_per_match
    FROM (
        SELECT match_id, COUNT(*) AS break_games_count
        FROM (
            SELECT match_id, game_id
            FROM game_point_by_point
            WHERE serving != scoring
            GROUP BY match_id, game_id
        )
        GROUP BY match_id
    )
"""
df = con.execute(query).df()
df
```
