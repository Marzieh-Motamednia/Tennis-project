<h3 style="color:#176B87;">نکات مهم ساختار:</h3>
<ul>
  <li><b>data/:</b> محل قرارگیری فایل‌های دیتابیس، خروجی‌های کوئری و فایل‌های کمکی.</li>
  <li><b>questions/:</b> تمام فعالیت‌های اصلی پروژه اینجا انجام می‌شود. هر پوشه شماره‌دار (مثل 001) متعلق به سوال مربوطه است.</li>
  <li><b>new-questions/:</b> محل ثبت سوالات جدیدی که در طول مسیر به ذهن اعضای تیم می‌رسد.</li>
</ul>

<h2 style="color:#1d7874;">📝 مدیریت وضعیت (status.txt)</h2>
<p>برای اینکه پیشرفت کار برای همه شفاف باشد، فایل <code>questions/status.txt</code> باید همیشه آپدیت باشد.</p>
<ul style="list-style: none;">
  <li>❌ <b>انجام نشده:</b> سوال هنوز شروع نشده است.</li>
  <li>⏳ <b>در حال انجام:</b> یکی از اعضا مسئولیت را پذیرفته و در حال تحلیل است.</li>
  <li>✅ <b>انجام شده:</b> تحلیل نهایی شده و فایل‌ها در پوشه مربوطه موجود هستند.</li>
</ul>

<div dir="rtl" style="font-family: Tahoma, sans-serif; background: #000000; padding: 20px; border-radius: 15px; margin-top: 20px; border: 1px solid #feb2b2;">
<h2 style="color:#c53030;">🚀 گردش کار تیمی (Git Workflow)</h2>
<p>برای جلوگیری از Conflict (تداخل کد)، این چرخه را هر روز رعایت کنید:</p>
<ol>
  <li><b>قبل از شروع کار:</b> دریافت آخرین تغییرات تیم: <br><code>git pull</code></li>
  <li><b>انجام کار:</b> اعمال تغییرات روی فایل‌ها در شاخه (Branch) مخصوص خودتان.</li>
  <li><b>بررسی تغییرات:</b> <br><code>git status</code></li>
  <li><b>ذخیره تغییرات:</b> <br><code>git add .</code> <br> <code>git commit -m "توضیح کوتاه"</code></li>
  <li><b>ارسال به سرور:</b> <br><code>git push</code></li>
</ol>
